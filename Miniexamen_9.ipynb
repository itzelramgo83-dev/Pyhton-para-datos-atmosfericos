```python
# d)

s4 = pd.Series([1, 2, 3], index=["a", "b", "c"])
s5 = pd.Series([10, 20, 30], index=["b", "c", "d"])
print(s4 + s5)
# Las etiquetas que no coinciden dan NaN, porque no hay con quien sumarlas. Asi que pandas hace coincidir los datos de ambas Series segun
# sus etiquetas (no segun su posicion, donde una etiqueta existe en ambas se opera, y donde falta en alguna el resultado es NaN (dato
# faltante). El index del resultado es la union de los dos index.
```

```python
# 1.3 Funciones de numpy sobre Series

#a)
print(np.sqrt(s1))
print(np.exp(s1))
print(np.log(s1))

#b)
print(np.log(s2))
# Regresa con Series (no un arreglo de numpy) y conservan el MISMO index de la Series original.

#c)Porque las Series de pandas no tienen metodos propios, ya estan en numpy. numpy hace el calculo y pandas devuelve el resultado en
# una Series, conservando el index.
```

```python
# 1.4 Estadstica básica

#a) Calcular:

print(s1.mean())
print(s1.sum())
print(s1.std())
print(s1.min())
print(s1.max())
```

```python
# b) Regresa una Series con un resumen estadistico
s1.describe()

```

```python
#c) NO dan lo mismo,  ddof  es el numero que se resta de N en el denominador: la desviacion se calcula con N - ddof.
# Ppandas usa ddof=1 por defecto (desviacion estandar MUESTRAL, divide entre N-1).
# Numpy usa ddof=0 por defecto (desviacion POBLACIONAL, divide entre N).
print(s1.std())
print(np.std(s1.values))
print(s1.std(ddof=0))          # coincide con numpy
print(np.std(s1.values, ddof=1))  # coincide con pandas
```

```python
# ---------- 2.4 Filtros (mascaras booleanas) ----------
# a)
mascara = tips['total_bill'] > 20
print(mascara)
print(tips[mascara])
```

```python
# b) Con las tres condiciones a la vez queda una sola fila
print(tips[(tips['tip'] > 5) & (tips['day'] == 'Sun') & (tips['size'] > 4)])
 
```

```python
# c)
print(tips[(tips['total_bill'] > 30) & (tips['tip'] > 5)])
 
```

```python
# d)
print(tips[(tips['day'] == 'Sun') | (tips['day'] == 'Sat')])
print(tips[tips['day'].isin(['Sun', 'Sat'])])
 
```

```python
# e)
print(tips.query('total_bill > 20'))
```

```python
# ---------- 2.5 Valores unicos y conteos ----------
# a)
print(tips['day'].unique())
print(tips['day'].nunique())
print(tips['day'].value_counts())
 
```

```python
# b)
for col in ['time', 'sex']:
    print(tips[col].unique())
    print(tips[col].nunique())
    print(tips[col].value_counts())
```

```python
# ---------- 2.6 Operaciones y nuevas columnas ----------
# a)
tips['tip_pct'] = tips['tip'] / tips['total_bill'] * 100
 
```

```python
# b)
tips['tip_per_person'] = tips['tip'] / tips['size']
 
```

```python
# c)
tips_nuevo = tips.assign(propina_grande=tips['tip'] > 5)
print(tips_nuevo)
print(tips)
# Diferencia: en (a) y (b) se MODIFICA tips directamente. assign NO modifica el original devuelve un
# DataFrame nuevo con la columna extra.Por eso assign sirve para encadenar pasos sin alterar los datos originales.
```

```python
# d) (no modifica tips; solo devuelve el nuevo DataFrame)
print(tips.assign(ratio=lambda x: x['tip'] / x['total_bill']))
```

```python
# e)
resultado = (tips
    .query('total_bill > 20')
    .assign(ratio=lambda x: x['tip'] / x['total_bill'])
    .assign(buena_propina=lambda x: x['ratio'] > 0.15)
)
print(resultado.head())

# Sin lambda:
paso1 = tips.query('total_bill > 20')
paso2 = paso1.assign(ratio=paso1['tip'] / paso1['total_bill'])
resultado_sin_lambda = paso2.assign(buena_propina=paso2['ratio'] > 0.15)
print(resultado_sin_lambda.head())
# Se necesitan 3 pasos y 3 variables, es decir, 2 intermedias. La lambda recibe como x "el
# DataFrame tal como esta en ese punto de la cadena", asi no hacen falta variables intermedias.
```

```python
# ---------- 2.7 apply y funciones ----------
# a)
def clasificar(x):
    if x < 2:
        return 'baja'
    elif x < 5:
        return 'media'
    else:
        return 'alta'
 
tips['categoria_propina'] = tips['tip'].apply(clasificar)
 
```

```python
# b)
tips['categoria_propina'] = tips['tip'].apply(
    lambda x: 'baja' if x < 2 else ('media' if x < 5 else 'alta'))
 
```

```python
# c) axis=1: la funcion recibe cada fila completa.
tips['tipo_cuenta'] = tips.apply(
    lambda fila: 'caro' if (fila['total_bill'] > 30 and fila['size'] > 3) else 'normal',
    axis=1)
print(tips.head())
```

```python
# ---------- 2.8 Ordenar ----------
# a)
print(tips.sort_values(by='tip', ascending=False))
 
```

```python
# b) Primero por 'day' y, dentro de cada dia, por 'tip'.
print(tips.sort_values(by=['day', 'tip']))
 
```

```python
# c) sort_index() reordena las filas por su index. Con tips no se notaria
# porque su index ya esta en orden ascendente, ordenar por index no cambia nada.
print(tips_ordenado.sort_index())
print(tips.sort_index())
 
```

```python
# ---------- 2.9 groupby ----------
g = tips.groupby('day', observed=True)
print(g)
print(g.groups)
print(g.size())
print(g.get_group('Sun'))
for nombre, grupo in g:
    print(f'--- {nombre} ---')
    print(grupo.head(2))
 
# a)
print(tips.groupby('day', observed=True)['tip'].mean())
 
```

```python
# b)
print(tips.groupby('sex', observed=True)['total_bill'].mean())
 
```

```python
# c)
print(tips.groupby('day', observed=True)['tip'].agg(['mean', 'std', 'count']))
 
```

```python
# d) El index del resultado es un MultiIndex de dos niveles (day, time): cada fila se identifica por la pareja (dia, momento).
res_d = tips.groupby(['day', 'time'], observed=True)['tip'].mean()
print(res_d)
print(res_d.index)
```

```python
# e)
print(tips.groupby('day', observed=True).agg(
    {'tip': 'mean', 'total_bill': 'sum', 'size': 'max'}))
```

```python
# ---------- 2.10 Pivot tables ----------
# a)
pivot_a = tips.pivot_table(index='day', columns='time', values='tip',
                           aggfunc='mean', observed=True)
print(pivot_a)
 
```

```python
# b)
pivot_b = tips.pivot_table(index='sex', columns='smoker',
                           values='total_bill', aggfunc='sum', observed=True)
print(pivot_b)
 
```

```python
# c) Ambos agrupan y agregan; la diferencia esta en la FORMA del resultado. groupby(['day','time']) da formato "largo", pivot_table da formato "ancho"
# combinaciones que no existen aparecen como NaN ( como sabado-almuerzo). En esencia, pivot_table equivale a un groupby seguido de .unstack() sobre uno de los niveles.
print(res_d.unstack())
```
