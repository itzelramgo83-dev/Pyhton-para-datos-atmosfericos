```python
from pathlib import Path
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
```

```python
ds = xr.open_dataset(Path.home() / "Downloads" / "gfs_data.nc")
t_2 = ds["t"].sel(lon=100, lev=2)
# una columna de la malla, nivel 2
t_10 = ds["t"].sel(lon=100, lev=10) # misma columna, nivel 10
lat = t_10["geolat"]
# igual para todos los niveles
```

```python
# 2a-b) Anatomía de una figura
fig, ax = plt.subplots()
ax.plot(lat, t_10)
ax.set_xlabel('Latitud [°]')
ax.set_ylabel('Temperatura [K]')
ax.set_title('Temperatura en lev=10, lon=100')
plt.show()
```

```python
#2c) Dos curvas + leyenda
fig, ax = plt.subplots()
ax.plot(lat, t_2, label='lev=2')
ax.plot(lat, t_10, label='lev=10')
ax.set_xlabel('Latitud [°]')
ax.set_ylabel('Temperatura [K]')
ax.set_title('Temperatura por nivel')
ax.legend()
plt.show()
```

```python
# 2d) Posiciones de la leyenda (tres o más)
for loc in ['upper left', 'lower right', 'upper center']:
    fig, ax = plt.subplots()
    ax.plot(lat, t_2, label='lev=2')
    ax.plot(lat, t_10, label='lev=10')
    ax.legend(loc=loc)
    ax.set_title(f"loc='{loc}'")
    plt.show()
```

```python
# 2e) Leyenda fuera de la gráfica + otras opciones de la tabla
fig, ax = plt.subplots()
ax.plot(lat, t_2, label='lev=2')
ax.plot(lat, t_10, label='lev=10')
ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1))
plt.show()
 
fig, ax = plt.subplots()
ax.plot(lat, t_2, label='lev=2')
ax.plot(lat, t_10, label='lev=10')
ax.legend(loc='lower center', bbox_to_anchor=(0.5, 1.02), ncol=2,
          title='Nivel', frameon=False, fontsize='small')
plt.show()
```

```python
# 4h) twinx vs secondary_yaxis
# Twinx crea un segundo eje que comparte x pero grafica OTROS datos con su propia escala y.
# Secondary_yaxis no grafica datos: solo muestra los MISMOS datos en otra escala/unidad,
#    definida por una función de conversión (ida y vuelta).
 
```

```python
# 5. Datos de la sección
col = dict(lat=100, lon=100)
t_col = ds['t'].sel(**col).values[1:]
zh = ds['zh'].sel(**col).values
z_km = 0.5 * (zh[:-1] + zh[1:])[1:] / 1000
delp = ds['delp'].sel(**col)
p_hpa = (delp.cumsum('lev') - delp / 2).values[1:] / 100
 
# 5a) Temperatura vs altura
fig, ax = plt.subplots()
ax.plot(t_col, z_km)
ax.set_xlabel('Temperatura [K]')
ax.set_ylabel('Altura [km]')
plt.show()
# en un perfil vertical la altura es la variable independiente física y
# "arriba" debe verse arriba; por convención meteorológica la variable
# medida va en x y la coordenada vertical en y.
```

```python
# 5b) Presión y eje invertido
fig, ax = plt.subplots()
ax.plot(t_col, p_hpa)
ax.set_xlabel('Temperatura [K]')
ax.set_ylabel('Presión [hPa]')
ax.invert_yaxis()
plt.show()
```

```python
#  5c) Escala logarítmica con marcas
fig, ax = plt.subplots()
ax.plot(t_col, p_hpa)
ax.set_yscale('log')
vals = [1000, 500, 200, 100, 50, 10, 1]
ax.set_yticks(vals, [str(v) for v in vals])
ax.invert_yaxis()
ax.set_xlabel('Temperatura [K]')
ax.set_ylabel('Presión [hPa]')
plt.show()
 
```

```python
# 5d) axhline con etiqueta
fig, ax = plt.subplots()
ax.plot(t_col, p_hpa, label='Perfil')
ax.axhline(100, color='r', linestyle='--', label='100 hPa')
ax.set_yscale('log')
ax.set_yticks(vals, [str(v) for v in vals])
ax.invert_yaxis()
ax.set_xlabel('Temperatura [K]')
ax.set_ylabel('Presión [hPa]')
ax.legend()
plt.show()
```

```python
#  5e) Cinco columnas superpuestas
fig, ax = plt.subplots()
for la, lo in [(20, 50), (100, 100), (150, 30), (200, 200), (60, 180)]:
    col = dict(lat=la, lon=lo)
    t_col = ds['t'].sel(**col).values[1:]
    delp = ds['delp'].sel(**col)
    p_hpa = (delp.cumsum('lev') - delp / 2).values[1:] / 100
    ax.plot(t_col, p_hpa, alpha=0.6, label=f'lat={la}, lon={lo}')
ax.set_yscale('log')
ax.set_yticks(vals, [str(v) for v in vals])
ax.invert_yaxis()
ax.set_xlabel('Temperatura [K]')
ax.set_ylabel('Presión [hPa]')
ax.legend(fontsize='small')
plt.show()
```

```python
#  6a) fill_between entre t_2 y t_10
fig, ax = plt.subplots()
ax.plot(lat, t_2, label='lev=2')
ax.plot(lat, t_10, label='lev=10')
ax.fill_between(lat, t_2, t_10, alpha=0.3)
ax.legend()
plt.show()
```

```python
# %% 6b) Anomalía: where, con y sin interpolate
anom = t_10 - t_10.mean()
fig, axs = plt.subplots(1, 2, figsize=(11, 4), sharey=True)
for ax, interp in zip(axs, [False, True]):
    ax.plot(lat, anom, color='k', linewidth=0.8)
    ax.axhline(0, color='k', linewidth=0.5)
    ax.fill_between(lat, anom, 0, where=anom > 0, color='tab:red',
                    alpha=0.5, interpolate=interp)
    ax.fill_between(lat, anom, 0, where=anom < 0, color='tab:blue',
                    alpha=0.5, interpolate=interp)
    ax.set_title(f'interpolate={interp}')
plt.show()
# R: sin interpolate, el sombreado se corta en el último punto de datos antes
# del cruce y quedan huecos en los cruces por cero; con interpolate=True el
# relleno llega hasta el punto exacto donde la curva cruza el cero.
 
```

```python
# %% 6c-d) Perfil medio y banda mín-máx
tm = ds['t'].mean(['lat', 'lon'])
tmin = ds['t'].min(['lat', 'lon'])
tmax = ds['t'].max(['lat', 'lon'])
lev = ds['t'].lev
fig, ax = plt.subplots()
ax.plot(tm, lev, color='k', label='Promedio')
ax.fill_betweenx(lev, tmin, tmax, alpha=0.3, label='Mín–máx')
ax.invert_yaxis()
ax.set_xlabel('Temperatura [K]')
ax.set_ylabel('Nivel (lev)')
ax.legend()
plt.show()
```
